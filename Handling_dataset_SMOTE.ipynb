#### SMOTE:
it is called as synthetic minority oversampling technique which is a machine learning technique used to generate the synthetic instances for Minority class which has significantly fewer instance as compare with majority class, it generates the synthetic instances for minority class by interpolating between existing instances, 

```python
from sklearn.datasets import make_classification
import pandas as pd
```

```python
x,y=make_classification(n_samples=1000,n_clusters_per_class=1,n_features=2,n_redundant=0,weights=[0.90],random_state=12)
```

```python
df1=pd.DataFrame(x,columns=['f1','f2'])
df2=pd.DataFrame(y,columns=['target'])
df=pd.concat([df1,df2],axis=1)
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['f1'],df['f2'],c=df['target'])
```

## As we can see that All Yellow colour Minority class Instances are significantly fewer as compare with Majority class 

```python
## Performing oversampling using Smote
from imblearn.over_sampling import SMOTE
oversample=SMOTE()
x,y=oversample.fit_resample(df[['f1','f2']],df['target'])
```

```python
y.shape
```

```python
df1=pd.DataFrame(x,columns=['f1','f2'])
df2=pd.DataFrame(y,columns=['target'])
df_oversampled=pd.concat([df1,df2],axis=1)
```

```python
plt.scatter(df_oversampled['f1'],df_oversampled['f2'],c=df_oversampled['target'])
```

## As we can see that All the Minority class Instances oversampled by Interpolating between Existed instances
