#### TARGET GUIDED ORDINAL ENCODING
In this technique we encode the categorical variable based on it's relationship with Target Variable. This encoding technique is very useful when categorical variable contains large number of unique values and we want to use this categorical variable as feature in my machine learning model.

in this encoding technique we replace each category of the categorical variable into numeric value based on the mean or median of the target varibale with respect to that particular category.This creates the Monotonic realtionship between Categorical Variable and Target Variable which somehow improves the Predictive Power of my machine learning Model.

```python
import pandas as pd
```

```python
import pandas as pd
##create a sample DataFrame with a categorical Variable and a target variable
df=pd.DataFrame(
    {
        'city':['New York','London','Paris','Tokyo','New York','Paris'],
        'price':[200,150,300,250,180,320]
    }
)
```

```python
df.head()
```

```python
mean_value=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_value
```

```python
df['target_encode']=df['city'].map(mean_value)
```
