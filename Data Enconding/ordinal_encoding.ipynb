### ORDINAL ENCODING:
In this technique we encode the categorical data into numeric values , each category encoded based on it's position or order in the categorical variable 


```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'Size':['small','medium','large','Extra_Large']
})  
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=[['small','medium','large','Extra_Large']])
ordinal_encoded=oe.fit_transform(df[['Size']])
```

```python
#Here we can see that based on the order and position all the categories encoded 
ordinal_encoded
```

```python
ordinal_df=pd.DataFrame(ordinal_encoded,columns=oe.get_feature_names_out())
```

```python
pd.concat([ordinal_df,df],axis=1)
```
