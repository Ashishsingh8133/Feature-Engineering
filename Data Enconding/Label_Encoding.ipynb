### LABLE ENCODING:
In this encoding technique we encode each category of the Categorical variable with unique numeric values and these values will be assigned to each category according to alphabetical order such as 'Red,Blue' then as B comes first then it will get numerical value as '0' and in 'Red' 'R' comes after 'B' it will get numerical value as '1', But in this encoding we might get a problem is that our machine learning model get confused it may compare like RED>BLUE because of encoded values , it should not happend, we don't want ranking in this case ' TO rank the categories we use different encoding technique called as 'Ordinal Encoding'

```python
import pandas as pd
```

```python
##create a Dataframe
df=pd.DataFrame({
    'color':['red','blue','green']
})
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
label_encoded=le.fit_transform(df['color'])
```

```python
### here we can see in the array as Blue got '0', Green got '1' and Red got '2'

label_encoded
```
