### Nominal Encoding/OpenHotEncoding:
In this encoding technique we encode the category of the categorical variable into numeric values 0's and 1's , but there is issue with such type of encoding is that ,if  a categorical variable contains 100 unique categories then after performing OneHotEncoding it will add 100 new features into the dataset which creates 'SparseMatrix' because of this sparse matrix my machine learning model will got the issue of 'OVERFITTING' means it will learn each and every pattern of the trained dataset and it will not perform well on new data.

```python
import pandas as pd
import seaborn as sns

```

```python
##create a Dataframe
df=pd.DataFrame({
    'color':['red','blue','green']
})
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
df_encoded=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder.transform([['red']]).toarray()
```

```python
df_encoded
```

```python
pd.concat([df,df_encoded],axis=1)
```
