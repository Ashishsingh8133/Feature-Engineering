Imputation techniques to deal with missing values instead of directly deleting the rows and columns
1. Mean Imputation technique

```python
df.head()
```

```python
sns.histplot(df['age'],kde=True,color='teal')
```

```python
df['Age_Mean']=df['age'].fillna(df['age'].mean())
```

```python
df.head()
```

```python
df.drop(columns='Age_Median',inplace=True)
```

```python
df.head()
```

```python
df[['Age_Mean','age']]
```

2. Median Imputation Techniques 
we used it when we have outliers in the datset then we will fill the missing values with median

```python
df['Age_Median']=df['age'].fillna(df['age'].median())
```

```python
df.head()
```

```python
df[['Age_Median','Age_Mean','age']]
```

3. Mode Imputation Technique-- it can be used for Categorical Values.

```python
df[df['embarked'].isnull()]
```

```python
df['embarked'].unique()
```

```python
mode_value=df[df['embarked'].notna()]['embarked'].mode()[0]
```

```python
df['embarked_mode']=df['embarked'].fillna(mode_value)
```
